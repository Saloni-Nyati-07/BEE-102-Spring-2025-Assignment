```python
import numpy as np
import pandas as pd

# Example: define your sequence here
sequence = "ACGTGCTAGCTAGCTAGGCTA"

# Define nucleotides
nucleotides = ['A', 'C', 'G', 'T']

# Initialize the transition count matrix
transition_counts = pd.DataFrame(0, index=nucleotides, columns=nucleotides)

# Count transitions
for i in range(len(sequence) - 1):
    curr_nt = sequence[i]
    next_nt = sequence[i + 1]
    if curr_nt in nucleotides and next_nt in nucleotides:
        transition_counts.loc[curr_nt, next_nt] += 1

# Normalize to get the transition probabilities
transition_matrix = transition_counts.div(transition_counts.sum(axis=1), axis=0)

print("Transition Matrix (First-order Markov Model):")
print(transition_matrix.fillna(0))
```
